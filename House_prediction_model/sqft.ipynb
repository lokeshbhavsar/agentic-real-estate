```python
import numpy as np
import pandas as pd
df=pd.read_csv('gurgaon.csv',usecols=[
'MAP_DETAILS',
'CARPET_SQFT',
"PREFERENCE",
"PROPERTY_TYPE",
'MAX_PRICE',
'SUPERBUILTUP_SQFT',
'BEDROOM_NUM', 
'BATHROOM_NUM',
'BALCONY_NUM',
'FACING', 'AGE',
"MAP_DETAILS",
"LOCALITY_WO_CITY"
])
#PRICE_SQFT will be no more effective as we have CARPET_SQFT
#Property Price=Carpet Area × Price Per Sq Ft
```

```python
df.columns
```

```python
df.info()
```

```python
df=df[df["PREFERENCE"]=='S']
df.head()
```

```python

df=pd.get_dummies(df,columns=['PROPERTY_TYPE'],dtype=int)
df
```

```python
df['SECTOR_NUMBER'] = df['LOCALITY_WO_CITY'].str.extract(r'(\d+)').astype(float).astype('Int64')
```

```python
df.head()
```

```python
df.drop(columns=['PREFERENCE','LOCALITY_WO_CITY'],inplace=True)
df
```

```python
import ast

```

```python
df['MAP_DETAILS'] = df['MAP_DETAILS'].apply(ast.literal_eval)
```

```python
df['LATITUDE'] = df['MAP_DETAILS'].apply(
    lambda x: float(x['LATITUDE'])
)

df['LONGITUDE'] = df['MAP_DETAILS'].apply(
    lambda x: float(x['LONGITUDE'])
)
```

```python
df.drop(columns=['MAP_DETAILS'],inplace=True)
```

```python
df
```

```python
df = df.dropna()
df.info()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['MAX_PRICE']),df['MAX_PRICE'],test_size=0.2,random_state=42)

X_train.iloc[0]
```

```python
y_train.iloc[0]
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(random_state=42)
model.fit(X_train,y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error

mse = mean_absolute_error(y_test,y_pred)

r2 = r2_score(y_test,y_pred)
print("mse",mse)
print("r2",r2)







```

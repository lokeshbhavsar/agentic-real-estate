```python
import numpy as np
import pandas as pd
data=pd.read_csv('gurgaon.csv')
data.head()
```

```python
import numpy as np
import pandas as pd
df=pd.read_csv('gurgaon.csv',usecols=[
'MAP_DETAILS',
'CARPET_SQFT',
"PREFERENCE",
"PROPERTY_TYPE",
'MAX_PRICE',
'SUPERBUILTUP_SQFT',
'BEDROOM_NUM', 
'BATHROOM_NUM',
'BALCONY_NUM',
'FACING', 'AGE',
"MAP_DETAILS",
"LOCALITY_WO_CITY"
])
#PRICE_SQFT will be no more effective as we have CARPET_SQFT
#Property Price=Carpet Area × Price Per Sq Ft
```

```python
df.columns
```

```python
df.info()
```

```python
df=df[df["PREFERENCE"]=='S']
df.head()
```

```python

df=pd.get_dummies(df,columns=['PROPERTY_TYPE'],dtype=int)
df
```

```python
df['SECTOR_NUMBER'] = df['LOCALITY_WO_CITY'].str.extract(r'(\d+)').astype(float).astype('Int64')
```

```python
df.head()
```

```python
df.drop(columns=['PREFERENCE','LOCALITY_WO_CITY'],inplace=True)
df
```

```python
import ast
```

```python
df['MAP_DETAILS'] = df['MAP_DETAILS'].apply(ast.literal_eval)
```

```python
df['LATITUDE'] = df['MAP_DETAILS'].apply(
    lambda x: float(x['LATITUDE'])
)

df['LONGITUDE'] = df['MAP_DETAILS'].apply(
    lambda x: float(x['LONGITUDE'])
)
```

```python
df.drop(columns=['MAP_DETAILS'],inplace=True)
```

```python
df
```

```python
df = df.dropna()
df.info()
```

```python
features = [
    'BEDROOM_NUM',
    'BATHROOM_NUM',
    'BALCONY_NUM',
    'FACING',
    'AGE',
    'CARPET_SQFT',
    'SUPERBUILTUP_SQFT',
    'PROPERTY_TYPE_Independent House/Villa',
    'PROPERTY_TYPE_Independent/Builder Floor',
    'PROPERTY_TYPE_Residential Apartment',
    'PROPERTY_TYPE_Residential Land',
    'SECTOR_NUMBER',
    'LATITUDE',
    'LONGITUDE'
]

X = df[features]


```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.neighbors import NearestNeighbors
lat_idx = X.columns.get_loc('LATITUDE')
lon_idx = X.columns.get_loc('LONGITUDE')

# APPLY WEIGHT AFTER SCALING
X_scaled[:, lat_idx] *= 10
X_scaled[:, lon_idx] *= 10

knn = NearestNeighbors(
    n_neighbors=10,
    metric='euclidean'
)

knn.fit(X_scaled)
```

```python
selected_property = X_scaled[100].reshape(1, -1)
print(X_scaled[100])
print(selected_property)

distances, indices = knn.kneighbors(selected_property)

print(indices)
print(distances)
```

```python
recommended = data.loc[df.index[indices[0]]]

print(recommended)
```

```python
from joblib import dump
artifact = {
    "knn": knn,
    "scaler": scaler,
    "features": features,
    "training_index": df.index.tolist(),
    "data": data.loc[df.index].reset_index(drop=True)
}

dump(artifact, "recommendation.pkl")
```
